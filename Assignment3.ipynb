**Why few-shot helps (or does not)?**

Why Few-Shot Helps

Few-shot prompting improves performance because it provides examples of how ambiguous complaints should be classified.

Benefits:

Clarifies category boundaries:
Example: "No service after inserting SIM" could be Device or Network. Examples show that the main customer impact determines classification.
Reduces inconsistent decisions:
The model learns the expected classification style.
Handles edge cases better:
Examples teach the model how to resolve overlapping categories.
Improves alignment with company-specific rules:
Different telecom companies may define categories differently.
**3(b). When Few-Shot Breaks?**

Few-shot prompting can fail when:

1. Poor examples are provided

If examples contain incorrect classifications, the model may learn wrong patterns.

2. Too few examples

A small number of examples may not cover complex complaint types.

3. Overfitting to examples

The model may copy examples instead of understanding the complaint.

Example:

Example says "slow internet → Network"
Model incorrectly classifies all speed-related complaints as Network even when the device is causing the issue.
4. New unseen cases

If a complaint is very different from the examples, the model may struggle.

5. Conflicting examples

Examples with inconsistent labels can confuse the model.

4. Technique Comparison
Approach	Strength	Weakness
Zero-Shot	Simple, flexible, requires no examples	May struggle with edge cases
Few-Shot	Better consistency and handles ambiguity	Depends heavily on example quality
Final Recommendation

For telecom complaint classification with overlapping categories, Few-Shot prompting is preferred because edge cases require learning classification patterns.

However, the examples must be carefully selected, especially examples showing boundary cases between Billing, Network, and Device. A hybrid approach (few-shot + clear classification rules) would provide the most reliable results.
